```python
import tensorflow as tf
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import numpy as np
import time
```

```python
# ==========================================
# 1. Data Loading & Preprocessing
# ==========================================
print("Loading Data...")
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize pixel values (0 to 1)
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# Flatten the images for Logistic Regression and Decision Tree
# (60000, 28, 28) -> (60000, 784)
x_train_flat = x_train.reshape((-1, 784))
x_test_flat = x_test.reshape((-1, 784))

results = {}
```

```python
print(x_train.shape)
```

# 2. Decision Tree (The Baseline)

```python
print("\n--- Training Decision Tree ---")
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(x_train_flat, y_train)

ytrainPred = dt_model.predict(x_train_flat)
ytestPred = dt_model.predict(x_test_flat)

train_acc = accuracy_score(y_train, ytrainPred)
test_acc = accuracy_score(y_test, ytestPred)

print(f"Decision Tree Accuracy, Train accuracy -" , train_acc)
print(f"Decision Tree Accuracy, Test accuracy -" , test_acc)
```

# 2. Logistic Regression

```python
print("\n--- Training Logistic Regression ---")
# Increased max_iter is often necessary for convergence with flattened data
lr_model = LogisticRegression(random_state=42, max_iter=1000)
lr_model.fit(x_train_flat, y_train)

ytrainPred = lr_model.predict(x_train_flat)
ytestPred = lr_model.predict(x_test_flat)

train_acc = accuracy_score(y_train, ytrainPred)
test_acc = accuracy_score(y_test, ytestPred)

print(f"Logistic Regression Accuracy, Train accuracy: {train_acc:.4f}")
print(f"Logistic Regression Accuracy, Test accuracy: {test_acc:.4f}")
```

# 3. Multi layer perceptron

```python
from tensorflow.keras import layers, models

# 1. Initialize the model
model = models.Sequential()

# 2. Add the Input/Flatten layer (converts 28x28 images into a 784-pixel vector)
model.add(layers.Flatten(input_shape=(28, 28)))

# 3. Add the Hidden layer (256 neurons with ReLU activation)
model.add(layers.Dense(256, activation='relu'))

# 4. Add Dropout (helps prevent overfitting by randomly turning off 20% of neurons)
model.add(layers.Dropout(0.2))

# 5. Add the Output layer (10 neurons for 10 classes, using Softmax for probability)
model.add(layers.Dense(10, activation='softmax'))

# 6. Compile the model
from keras.optimizers import SGD
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
from tensorflow.keras.utils import to_categorical

# Convert target labels to "One-Hot" encoding
y_train_encoded = to_categorical(y_train)
y_test_encoded = to_categorical(y_test)
```

```python
# 1. Train the model
model.fit(x_train, y_train_encoded, epochs=5, batch_size=64)

# 2. Evaluate on Train data (to check for overfitting)
train_loss, train_acc = model.evaluate(x_train, y_train_encoded)

# 3. Evaluate on Test data (the "final exam")
test_loss, test_acc = model.evaluate(x_test, y_test_encoded)

# 4. Print results
print(f"Train Accuracy: {train_acc * 100:.2f}%")
print(f"Test Accuracy:  {test_acc * 100:.2f}%")
```
